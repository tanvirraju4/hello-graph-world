```python
import networkx as nx
import matplotlib.pyplot as plt
```

## 1. Define Nodes and Edges
The Krackhardt Kite has 10 nodes and the following edges as shown in the assignment diagram.

```python
# Create undirected graph
G = nx.Graph()

# Add nodes
nodes = ["Andre", "Beverly", "Carol", "Diane", "Ed",
         "Fernando", "Garth", "Heather", "Ike", "Jane"]
G.add_nodes_from(nodes)

# Add edges (based on Krackhardt Kite structure)
edges = [
    ("Andre",    "Beverly"),
    ("Andre",    "Carol"),
    ("Andre",    "Diane"),
    ("Andre",    "Fernando"),
    ("Beverly",  "Diane"),
    ("Beverly",  "Ed"),
    ("Beverly",  "Garth"),
    ("Carol",    "Diane"),
    ("Carol",    "Fernando"),
    ("Diane",    "Fernando"),
    ("Diane",    "Garth"),
    ("Ed",       "Garth"),
    ("Fernando", "Garth"),
    ("Fernando", "Heather"),
    ("Garth",    "Heather"),
    ("Heather",  "Ike"),
    ("Ike",      "Jane"),
]
G.add_edges_from(edges)

print(f"Nodes ({G.number_of_nodes()}): {list(G.nodes)}")
print(f"Edges ({G.number_of_edges()}): {list(G.edges)}")
```

## 2. Visualise the Graph

```python
# Fixed positions to match the assignment diagram layout
pos = {
    "Carol":    (2, 4),
    "Andre":    (0, 3),
    "Fernando": (4, 3),
    "Diane":    (2, 2),
    "Beverly":  (0, 1),
    "Garth":    (4, 1),
    "Ed":       (2, 0),
    "Heather":  (6, 2),
    "Ike":      (8, 2),
    "Jane":     (10, 2),
}

plt.figure(figsize=(12, 6))

nx.draw_networkx_nodes(G, pos, node_color="#028090", node_size=800, alpha=0.9)
nx.draw_networkx_labels(G, pos, font_color="white", font_size=10, font_weight="bold")
nx.draw_networkx_edges(G, pos, edge_color="#0D1B3E", width=2, alpha=0.8)

plt.title("Krackhardt Kite Graph", fontsize=16, fontweight="bold", pad=20)
plt.axis("off")
plt.tight_layout()
plt.show()
```

## 3. Basic Graph Statistics

```python
print("=== Graph Summary ===")
print(f"Nodes          : {G.number_of_nodes()}")
print(f"Edges          : {G.number_of_edges()}")
print(f"Is connected   : {nx.is_connected(G)}")
print(f"Density        : {nx.density(G):.3f}")
print()
print("=== Degree per Node ===")
for node, degree in sorted(G.degree(), key=lambda x: -x[1]):
    print(f"  {node:<10} degree = {degree}")
print()
print("=== Centrality (Betweenness) ===")
betweenness = nx.betweenness_centrality(G)
for node, score in sorted(betweenness.items(), key=lambda x: -x[1]):
    print(f"  {node:<10} {score:.3f}")
```
